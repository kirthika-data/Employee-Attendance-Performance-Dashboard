```python
from google.colab import files
import pandas as pd
uploaded = files.upload()
filename = list(uploaded.keys())[0]
```

```python
raw = pd.read_excel(filename, header=None)
col = raw[1]

raw['date'] = raw[5].where(col == 'Attendance Date :').ffill()
df = raw[col.astype(str).str.strip().str.isdigit()].copy()

df = df.rename(columns={
    1: 'SNo', 2: 'E. Code', 3: 'Name', 5: 'Shift',
    6: 'S. InTime', 7: 'S. OutTime', 9: 'A. InTime', 10: 'A. OutTime',
    11: 'Work Dur.', 12: 'OT', 13: 'Tot. Dur.', 14: 'LateBy',
    15: 'EarlyGoingBy', 16: 'Status', 18: 'Punch Records',
    'date': 'Date'
})

df = df[['SNo', 'Date', 'E. Code', 'Name', 'Shift',
         'S. InTime', 'S. OutTime', 'A. InTime', 'A. OutTime',
         'Work Dur.', 'OT', 'Tot. Dur.', 'LateBy',
         'EarlyGoingBy', 'Status', 'Punch Records']]

df['Date'] = pd.to_datetime(df['Date'], format='%d-%b-%Y')
df['Weekday'] = df['Date'].dt.day_name()
df.loc[df['Weekday'].isin(['Saturday','Sunday']) & (df['Status']=='Absent'), 'Status'] = 'WeeklyOff'
df = df.drop(columns=['Weekday'])
df['Date'] = df['Date'].dt.strftime('%d-%m-%Y')

df['Name'] = df['Name'].str.strip()
df['Status'] = df['Status'].str.strip()
df = df.sort_values('Date').reset_index(drop=True)
df['SNo'] = range(1, len(df) + 1)
df.to_excel('DailyAttendance_Detailed_Cleaned.xlsx', index=False)
files.download('DailyAttendance_Detailed_Cleaned.xlsx')
```
